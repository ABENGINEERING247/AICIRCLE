```python
# Electrical Calculator
# A simple program for common electrical calculations


# Keep the calculator running until the user chooses to exit
while True:

    # Display the calculator menu
    print("\nElectrical Calculator")
    print("--------------------")
    print("1 - Calculate Voltage")
    print("2 - Calculate Current")
    print("3 - Calculate Resistance")
    print("4 - Calculate Power")
    print("5 - Calculate Energy")
    print("6 - Series Resistance")
    print("7 - Parallel Resistance")
    print("8 - Voltage Comparison")
    print("9 - Power Comparison")
    print("e - Exit")

    # Ask the user to select an option
    choice = input("Enter your choice: ")

    # Calculate voltage using Ohm's Law
    if choice == "1":
        current = float(input("Enter current in amperes (A): "))
        resistance = float(input("Enter resistance in ohms (Ω): "))

        voltage = current * resistance

        print("Voltage =", round(voltage, 2), "V")

    # Calculate current using Ohm's Law
    elif choice == "2":
        voltage = float(input("Enter voltage in volts (V): "))
        resistance = float(input("Enter resistance in ohms (Ω): "))

        if resistance == 0:
            print("Error: Resistance cannot be zero.")
        else:
            current = voltage / resistance
            print("Current =", round(current, 2), "A")

    # Calculate resistance using Ohm's Law
    elif choice == "3":
        voltage = float(input("Enter voltage in volts (V): "))
        current = float(input("Enter current in amperes (A): "))

        if current == 0:
            print("Error: Current cannot be zero.")
        else:
            resistance = voltage / current
            print("Resistance =", round(resistance, 2), "Ω")

    # Calculate electrical power
    elif choice == "4":
        voltage = float(input("Enter voltage in volts (V): "))
        current = float(input("Enter current in amperes (A): "))

        power = voltage * current

        print("Power =", round(power, 2), "W")

    # Calculate electrical energy
    elif choice == "5":
        power = float(input("Enter power in watts (W): "))
        time = float(input("Enter time in hours (h): "))

        energy = power * time

        print("Energy =", round(energy, 2), "Wh")

    # Calculate series resistance
    elif choice == "6":
        resistance_one = float(input("Enter the first resistance (Ω): "))
        resistance_two = float(input("Enter the second resistance (Ω): "))

        total_resistance = resistance_one + resistance_two

        print("Total series resistance =",
              round(total_resistance, 2), "Ω")

    # Calculate parallel resistance
    elif choice == "7":
        resistance_one = float(input("Enter the first resistance (Ω): "))
        resistance_two = float(input("Enter the second resistance (Ω): "))

        if resistance_one + resistance_two == 0:
            print("Error: The sum of the resistances cannot be zero.")
        else:
            total_resistance = (
                resistance_one * resistance_two
            ) / (resistance_one + resistance_two)

            print("Total parallel resistance =",
                  round(total_resistance, 2), "Ω")

    # Compare the voltage level
    elif choice == "8":
        voltage = float(input("Enter a voltage value (V): "))

        if voltage < 50:
            print("Low voltage")
        elif voltage <= 240:
            print("Medium voltage for this simple exercise")
        else:
            print("High voltage for this simple exercise")

    # Compare the power level
    elif choice == "9":
        power = float(input("Enter a power value (W): "))

        if power < 100:
            print("Low power")
        elif power <= 1000:
            print("Moderate power")
        else:
            print("High power")

    # Exit the calculator
    elif choice == "e":
        print("Thank you for using the Electrical Calculator.")
        break

    # Handle an invalid menu choice
    else:
        print("Invalid choice. Please select an option from the menu.")
```
